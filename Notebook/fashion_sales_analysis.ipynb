```python
# 1. Revenue by Product Category
# Bar chart
# Compare the total revenue generated by each fashion category.

import matplotlib.pyplot as plt
import seaborn as sns


category_revenue = (
    sales_data.groupby("category")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(10, 5))
sns.barplot(
    data=category_revenue,
    x="category",
    y="net_sales"
)

plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(category_revenue)


```

How to interpret it
- The tallest bar represents the category generating the most revenue.
- The shortest bar represents the category generating the least revenue.

Business Recommendations
- Give high-revenue categories appropriate stock levels and marketing support.
- Investigate low-revenue categories to see whether pricing, product quality, or customer demand needs attention.
- Compare revenue with profit before deciding which category deserves the most investment.

```python
# 2. Monthly Sales Trend
# Line chart
# Show how sales change over time.

import matplotlib.pyplot as plt
import seaborn as sns

sales_data["order_date"] = pd.to_datetime(sales_data["order_date"])

monthly_sales = (
    sales_data.groupby(sales_data["order_date"].dt.to_period("M"))["net_sales"]
    .sum()
    .reset_index()
)

monthly_sales["month"] = (
    monthly_sales["order_date"].dt.to_timestamp()
)

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x="month",
    y="net_sales",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(monthly_sales[["month", "net_sales"]])
```

How to interpret it
- An upward trend indicates increasing monthly revenue.
- A downward trend indicates declining revenue.
- Sharp peaks may indicate promotions, holidays, or seasonal demand.

Business Recommendations
- Plan inventory ahead of months with consistently high sales.
- Investigate months with sharp declines.
- Compare trends across multiple years where possible before concluding that a pattern is seasonal.

```python
# 3. Sales by Season
# Bar chart
# Compare revenue across different seasons.

import matplotlib.pyplot as plt
import seaborn as sns


season_sales = (
    sales_data.groupby("season")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=season_sales,
    x="season",
    y="net_sales"
)

plt.title("Sales by Season")
plt.xlabel("Season")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()

print(season_sales)
```

How to interpret it

The tallest bar identifies the season with the highest total revenue.

Business Recommendations
- Increase stock and marketing before strong sales seasons.
- Investigate weak seasons and consider targeted promotions.
- Consider whether differences reflect the length of each season, not just customer demand.

```python
# 4. Online vs Physical Store Performance
# Bar/column chart
# Compare sales performance across sales channels.

import matplotlib.pyplot as plt
import seaborn as sns


channel_sales = (
    sales_data.groupby("sales_channel")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(8, 5))
sns.barplot(
    data=channel_sales,
    x="sales_channel",
    y="net_sales"
)

plt.title("Online vs Physical Store Performance")
plt.xlabel("Sales Channel")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()

print(channel_sales)
```

How to interpret it

Compare the online and physical-store bars to identify which channel generates more revenue.

Business Recommendations
- If online sales lead, investigate improving the website, delivery experience, and digital advertising.
- If physical stores lead, investigate store location, customer service, and in-store promotions.
- Compare costs and profit margins before reallocating the marketing budget.

```python
# 5. Top 10 Products by Revenue
# Horizontal bar chart
# Identify the products generating the most revenue.

import matplotlib.pyplot as plt
import seaborn as sns


top_products = (
    sales_data.groupby("product_name")["net_sales"]
    .sum()
    .nlargest(10)
    .sort_values()
    .reset_index()
)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=top_products,
    x="net_sales",
    y="product_name"
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Revenue")
plt.ylabel("Product")
plt.tight_layout()
plt.show()

print(top_products.sort_values("net_sales", ascending=False))
```

How to interpret it

The longest bar represents the highest-revenue product among the top 10.

Business Recommendations
- Maintain adequate stock of popular revenue-generating products.
- Identify the features, prices, and categories associated with successful products.
- Check profit margins and stock turnover; a product with high revenue is not necessarily the most profitable.

```python
# 6. Category × Season Performance
# Heatmap
# Identify which product categories perform best during different seasons.

import matplotlib.pyplot as plt
import seaborn as sns


category_season = sales_data.pivot_table(
    index="category",
    columns="season",
    values="net_sales",
    aggfunc="sum",
    fill_value=0
)

plt.figure(figsize=(10, 6))
sns.heatmap(
    category_season,
    annot=True,
    fmt=".0f",
    cmap="YlOrRd"
)

plt.title("Category vs Season Performance")
plt.xlabel("Season")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()

print(category_season)
```

How to interpret it
- Find the highest-value cell to identify the strongest category-season combination.
- Compare each category across seasons to identify changes in demand.

Business Recommendations
- Promote categories that perform well during specific seasons.
- Prepare seasonal inventory in advance.
- Consider seasonal bundles or discounts for categories that consistently perform poorly during particular periods.

```python
# 7. Regional Sales Performance
# Bar chart
# Compare revenue generated across regions.

import matplotlib.pyplot as plt
import seaborn as sns


regional_sales = (
    sales_data.groupby("region")["net_sales"]
    .sum()
    .sort_values(ascending=False)
    .reset_index()
)

plt.figure(figsize=(9, 5))
sns.barplot(
    data=regional_sales,
    x="region",
    y="net_sales"
)

plt.title("Regional Sales Performance")
plt.xlabel("Region")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(regional_sales)
```

How to interpret it

The highest bar represents the region generating the most revenue.

Business Recommendations
- Investigate successful regions to understand what drives their performance.
- Examine lower-revenue regions for differences in customer demand, product availability, or distribution.
- Compare revenue per customer and sales volume before concluding that one region has better purchasing behavior.

```python
# 8. Discount vs Quantity Sold
# Scatter plot
# Investigate whether higher discounts are associated with increased sales volume.

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.stripplot(
    data=sales_data,
    x="discount",
    y="quantity",
    jitter=0.2,
    alpha=0.4
)

plt.title("Discount vs Quantity Sold")
plt.xlabel("Discount (%)")
plt.ylabel("Quantity Sold")
plt.tight_layout()
plt.show()

print(
    "Correlation:",
    sales_data[["discount", "quantity"]].corr().iloc[0, 1]
)
```

Insight
The correlation coefficient of 0.001657 indicates virtually no linear relationship between discount percentage and quantity sold in this dataset.

This means that higher discounts are not associated with a consistent increase or decrease in the quantity of products purchased.

What does this mean for the business?
The result suggests that discount percentage alone may not be a useful predictor of how many units customers purchase. Other factors, such as product category, price, season, and customer preferences, may influence purchasing quantities.

Business Recommendations
1. Avoid unnecessary discounts: Do not automatically increase discounts expecting customers to purchase more units.
2. Test promotional campaigns: Compare sales volume and profit before and during discount campaigns.
3. Investigate other factors: Analyze how product categories, seasons, and sales channels relate to quantity sold.
4. Protect profit margins: Evaluate whether discounts generate enough additional sales to justify the reduction in selling price.

Simple Insight & Recommendation

**Insight:** The correlation between discount percentage and quantity sold is 0.00167, indicating virtually no linear relationship between the two variables in the dataset.

**Business Recommendation:** The business should avoid relying solely on discounts to increase sales volume. It should investigate other factors influencing purchasing behavior and evaluate promotional campaigns based on their impact on both sales and profitability.

```python
# 9. Sales Distribution by Category
# Boxplot
# Compare the distribution of sales across product categories and identify potential outliers.

import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(10, 6))

sns.boxplot(
    data=sales_data,
    x="category",
    y="net_sales"
)

plt.title("Sales Distribution by Category")
plt.xlabel("Product Category")
plt.ylabel("Net Sales per Record")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

How to interpret it
- The line inside each box represents the median sale.
- The box represents the middle 50% of sales values.
- Points beyond the whiskers are potential outliers, not necessarily errors.
- A wider box indicates greater variation in sales values.

Business Recommendations
- Investigate unusually large transactions to understand whether they represent bulk purchases or data errors.
- Compare typical sales values across categories, not just total revenue.
- Consider category-specific pricing and promotional strategies.

```python
# 10. Popular Sizes and Colors
# Bar chart
# Identify the most frequently purchased sizes and colors.

import matplotlib.pyplot as plt
import seaborn as sns


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

size_counts = (
    sales_data["size"].dropna()
    .value_counts()
    .reset_index()
)
size_counts.columns = ["size", "purchases"]

sns.barplot(
    data=size_counts,
    x="size",
    y="purchases",
    ax=axes[0]
)
axes[0].set_title("Popular Product Sizes")
axes[0].set_xlabel("Size")
axes[0].set_ylabel("Number of Records")

color_counts = (
    sales_data["color"].dropna()
    .value_counts()
    .reset_index()
)
color_counts.columns = ["color", "purchases"]

sns.barplot(
    data=color_counts,
    x="color",
    y="purchases",
    ax=axes[1]
)
axes[1].set_title("Popular Product Colors")
axes[1].set_xlabel("Color")
axes[1].set_ylabel("Number of Records")
axes[1].tick_params(axis="x", rotation=45)

plt.tight_layout()
plt.show()

print("Popular sizes:")
print(size_counts)

print("\nPopular colors:")
print(color_counts)
```

How to interpret it
- The tallest size bar identifies the most frequently recorded size.
- The tallest color bar identifies the most frequently recorded color.

Business Recommendations
- Stock frequently purchased sizes and colors appropriately.
- Investigate slow-moving sizes and colors before reducing their stock.
- Compare these results within each product category because size and color preferences may differ between products
